```python
import pandas as pd
import numpy as np

DATA_PROCESSED = '/home/ronak/Documents/primetrade-assignment/data/processed'

daily   = pd.read_csv(f'{DATA_PROCESSED}/daily_metrics.csv')
traders = pd.read_csv(f'{DATA_PROCESSED}/trader_summary.csv')
merged  = pd.read_csv(f'{DATA_PROCESSED}/merged_data.csv')

fear  = daily[daily['Classification'] == 'Fear']
greed = daily[daily['Classification'] == 'Greed']

print("=" * 55)
print("PnL STATS")
print("=" * 55)
print(f"Fear  — mean PnL : {fear['daily_pnl'].mean():.4f}")
print(f"Greed — mean PnL : {greed['daily_pnl'].mean():.4f}")
print(f"Fear  — median   : {fear['daily_pnl'].median():.4f}")
print(f"Greed — median   : {greed['daily_pnl'].median():.4f}")
print(f"Fear  — std      : {fear['daily_pnl'].std():.4f}")
print(f"Greed — std      : {greed['daily_pnl'].std():.4f}")

print()
print("=" * 55)
print("WIN RATE")
print("=" * 55)
print(f"Fear  — win rate : {fear['win_rate'].mean()*100:.2f}%")
print(f"Greed — win rate : {greed['win_rate'].mean()*100:.2f}%")

print()
print("=" * 55)
print("BEHAVIOR")
print("=" * 55)
print(f"Fear  — avg trades/day : {fear['num_trades'].mean():.2f}")
print(f"Greed — avg trades/day : {greed['num_trades'].mean():.2f}")
print(f"Fear  — avg size USD   : {fear['avg_size_usd'].mean():.2f}")
print(f"Greed — avg size USD   : {greed['avg_size_usd'].mean():.2f}")
print(f"Fear  — buy/sell ratio : {fear['buy_sell_ratio'].mean():.4f}")
print(f"Greed — buy/sell ratio : {greed['buy_sell_ratio'].mean():.4f}")

print()
print("=" * 55)
print("SEGMENTS")
print("=" * 55)
seg = daily.merge(traders[['Account','size_segment','winner_segment','frequency_segment']], on='Account', how='left')
print(seg.groupby(['size_segment','Classification'])['daily_pnl'].mean().unstack().round(4))
print()
print(seg.groupby(['winner_segment','Classification'])['daily_pnl'].mean().unstack().round(4))
print()
print(seg.groupby(['frequency_segment','Classification'])['daily_pnl'].mean().unstack().round(4))

print()
print("=" * 55)
print("DRAWDOWN")
print("=" * 55)
print(f"Avg max drawdown : {traders['max_drawdown'].mean():.4f}")
print(f"Worst drawdown   : {traders['max_drawdown'].min():.4f}")

print()
print("=" * 55)
print("TOP COINS BY PnL")
print("=" * 55)
print(merged.groupby(['Coin','Classification'])['Closed PnL'].mean().unstack().round(4).head(10))
```
